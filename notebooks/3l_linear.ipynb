```python
import dezero as dz
import dezero.functions as F
import numpy as np
import matplotlib.pyplot as plt
```

```python
def linear_simple(x, W, b=None):
    x, W = dz.as_variables(x, W)
    t = F.matmul(x, W)
    if b is None:
        return t

    y = t + b
    t.data = None
    return y
```

```python
%matplotlib inline
np.random.seed(0)
x = np.random.rand(100, 1)
y = np.sin(2 * np.pi * x) + np.random.rand(100, 1)

plt.plot(x, y, 'o')
```
